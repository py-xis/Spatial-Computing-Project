# Filling out the Missing Values in the NDVI column

```python
blr_df.isnull().sum()
```

```python
pune_df.isnull().sum()
```

```python
def interpolate(date, series, window=30):
    current_val = series.loc[date]
    if not pd.isna(current_val):
        return current_val
    
    # Find non-NaN values before and after
    before = series.loc[date - pd.Timedelta(days=window): date - pd.Timedelta(days=1)].dropna()
    after = series.loc[date + pd.Timedelta(days=1): date + pd.Timedelta(days=window)].dropna()
    
    if not before.empty and not after.empty:
        # Linear interpolation
        t1 = before.index[-1]
        t2 = after.index[0]
        y1 = before.loc[t1]
        y2 = after.loc[t2]
        weight = (date - t1).days / (t2 - t1).days
        return y1 + (y2 - y1) * weight
    
    elif not before.empty:
        # Forward fill
        return before.iloc[-1]
    
    elif not after.empty:
        # Backward fill
        return after.iloc[0]
    
    else:
        # No data in the window
        return np.nan
```

```python
blr_df["NDVI"] = [interpolate(date, blr_df["NDVI"]) for date in blr_df.index]
```

### There are negative values for dew - point temperature which is not possible for the cities under consideration and are most probably due to artificats of satellite observations

## To deal, with it any value below 5 degree celsius is considered a NaN value and then interpolated.

```python
def mark_as_nan(df : pd.DataFrame, col : str, threshold : float = 5.0) -> None :
    """
    Mark values as NaN if they are less than the threshold
    :param df: DataFrame to be converted
    :param col: Name of the column
    :param threshold: Threshold value
    :return: Nothing
    """
    df[col] = df[col].where(df[col] >= threshold, np.nan)
```

```python
def plot_dpt(df : pd.DataFrame) -> None :
    """
    Plot the DPT column
    :param df: DataFrame to be plotted
    :return: Nothing
    """
    plt.figure(figsize=(12, 6))
    plt.plot(df.index, blr_df['DPT'], alpha=0.7)
    plt.title('Date vs DPT')
    plt.xlabel('Date')
    plt.ylabel('DPT')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
plot_dpt(blr_df)
```
