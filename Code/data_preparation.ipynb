### Dropping Duplicates

```python
def clean_dates(df):
    df['date'] = pd.to_datetime(df['date'], errors='coerce').dt.date
    df = df.dropna(subset=['date'])
    df = df.drop_duplicates(subset='date')
    return df
```

### We can see that there is a mismatch between the number of entries across different datasets. For that I shall take all the common entries based on date in the tables

```python
# Extract date sets
lst_dates = set(blr_lst['date']) & set(hyd_lst['date']) & set(pune_lst['date'])
dpt_dates = set(blr_dpt['date']) & set(hyd_dpt['date']) & set(pune_dpt['date'])
ap_dates = set(blr_ap['date']) & set(hyd_ap['date']) & set(pune_ap['date'])
wind_dates = set(blr_wind['date']) & set(hyd_wind['date']) & set(pune_wind['date'])
at_dates = set(blr_at['date']) & set(hyd_at['date']) & set(pune_at['date'])
```

```python
# Final common dates
common_dates = lst_dates & dpt_dates & ap_dates & wind_dates & at_dates
common_dates = sorted(list(common_dates))  # Ensure chronological order
```

### Calculating the magnitude of wind speed and direction

```python
def calculateWindSpeed(df : pd.DataFrame) -> pd.DataFrame:
    """
    Calculate wind speed from u and v components.
    """
    df['WS'] = np.sqrt(df['u_component_of_wind_10m']**2 + df['v_component_of_wind_10m']**2)
    return df
```

```python
def calculateWindDirection(df : pd.DataFrame) -> pd.DataFrame:
    """
    Calculate wind direction from u and v components.
    """
    df['WSD'] = np.arctan2(df['u_component_of_wind_10m'], df['v_component_of_wind_10m']) * 180 / np.pi
    df['WSD'] = (df['WSD'] + 360) % 360  # Normalize to [0, 360)
    return df
```

```python
blr_wind_daily = calculateWindDirection(blr_wind_daily)
pune_wind_daily = calculateWindDirection(pune_wind_daily)
hyd_wind_daily = calculateWindDirection(hyd_wind_daily)
```

## Converting the units to Celsisus, metres/second, pascal to hPa

```python
def lstCorrect(df : pd.DataFrame) -> pd.DataFrame:
    """
    Corrects the LST values based on the RH values.
    """
    df['LST'] = df['LST'] * 0.1
    return df
```

```python
def toCelsius(df : pd.DataFrame, col: str) -> pd.DataFrame:
    """
    Converts Kelvin to Celsius.
    """
    df[col] = df[col] - 273.15
    return df
```

```python
def paTohPa(df : pd.DataFrame, col: str) -> pd.DataFrame:
    """
    Converts Pa to hPa.
    """
    df[col] = df[col] / 100
    return df
```

```python
blr_df = lstCorrect(blr_df)
hyd_df = lstCorrect(hyd_df)
pune_df = lstCorrect(pune_df)
```

```python
blr_df = toCelsius(blr_df, 'DPT')
hyd_df = toCelsius(hyd_df, 'DPT')
pune_df = toCelsius(pune_df, 'DPT')
```

```python
blr_df = paTohPa(blr_df, 'AP')
pune_df = paTohPa(pune_df, 'AP')
hyd_df = paTohPa(hyd_df, 'AP')
```

# Calculating RH

```python
blr_at_daily = toCelsius(blr_at_daily, 'airTemp')
hyd_at_daily = toCelsius(hyd_at_daily, 'airTemp')
pune_at_daily = toCelsius(pune_at_daily, 'airTemp')
```

```python
blr_df['AT'] = blr_at_daily['airTemp']
hyd_df['AT'] = hyd_at_daily['airTemp']
pune_df['AT'] = pune_at_daily['airTemp']
```

```python
def compute_rh_from_air_temp(df, temp_col='T', dew_point_col='DP', rh_col='RH'):
    """
    Compute Relative Humidity and return the full DataFrame with a new RH column.

    Parameters:
    - df: pandas DataFrame
    - temp_col: temperature column name in °C
    - dew_point_col: dew point temperature column name in °C
    - rh_col: name for the RH column to be added

    Returns:
    - Modified DataFrame with an added RH column
    """
    T = df[temp_col]
    DP = df[dew_point_col]

    rh = ((112 - 0.1 * T + DP) / (112 + 0.9 * T)) * 100
    df[rh_col] = rh.clip(lower=0, upper=100)
    df.drop(columns = [temp_col], inplace=True)
    return df
```

```python
blr_df = compute_rh_from_air_temp(blr_df, temp_col='AT', dew_point_col='DPT')
hyd_df = compute_rh_from_air_temp(hyd_df, temp_col='AT', dew_point_col='DPT')
pune_df = compute_rh_from_air_temp(pune_df, temp_col='AT', dew_point_col='DPT')
```
