## Plotting a chart of time vs LST

```python
def plotTimeVsLST(df: pd.DataFrame, city : str) -> None:
    """
    Plots an interactive Time vs LST chart using Plotly, with years shown on the x-axis.
    
    :param df: DataFrame containing 'Date' and 'LST' columns.
    """
    # Ensure 'Date' is datetime
    df['Date'] = pd.to_datetime(df['Date'])
    
    # Add 'Year' column
    df['Year'] = df['Date'].dt.year
    
    # Plot using Plotly
    fig = px.line(
        df,
        x='Date',
        y='LST',
        title=f'Date vs Land Surface Temperature (LST) for {city}',
        labels={'Date': 'Year', 'LST': 'Land Surface Temperature (°C)'},
        template='plotly_white'
    )
    
    # Customize x-axis to only show year ticks
    fig.update_layout(
        xaxis=dict(
            tickmode='array',
            tickvals=pd.date_range(start=df['Date'].min(), end=df['Date'].max(), freq='YS'),
            tickformat='%Y',
            tickangle=45
        ),
        title_font_size=20,
        font=dict(size=14),
        margin=dict(t=50, b=50, l=40, r=40),
        hovermode="x unified"
    )
    
    fig.update_traces(line=dict(width=2), marker=dict(size=4))
    fig.show()
```

```python
plotTimeVsLST(blr_df, 'Bengaluru')
```

```python
plotTimeVsLST(pune_df, 'Pune')
```

```python
plotTimeVsLST(hyd_df, 'Hyderabad')
```

## Calculating the Pearson Correlation Coefficient and Visualizing it

```python
def compute_pearson_correlation(df: pd.DataFrame, predictors: list) -> pd.DataFrame:
    """
    Computes the Pearson correlation coefficient of all columns in the DataFrame
    against the specified predictor variable(s).

    :param df: A pandas DataFrame with all numeric columns.
    :param predictors: A list of column names to use as predictors.
    :return: A DataFrame of Pearson correlation coefficients (rows: features, columns: predictors)
    """
    # Ensure predictor columns are in the DataFrame
    missing = [col for col in predictors if col not in df.columns]
    if missing:
        raise ValueError(f"Predictor columns not found in DataFrame: {missing}")

    # Only numeric columns
    numeric_df = df.select_dtypes(include='number')

    # Drop predictors from input features if they exist in the numeric set
    input_vars = [col for col in numeric_df.columns if col not in predictors]

    # Compute correlation matrix
    correlations = numeric_df[input_vars].corrwith(numeric_df[predictors[0]]).to_frame(name=predictors[0])
    
    for pred in predictors[1:]:
        correlations[pred] = numeric_df[input_vars].corrwith(numeric_df[pred])
    
    correlations.index.name = 'Input_Variables'
    return correlations
```

```python
corr_coef_blr = compute_pearson_correlation(blr_df, ['LST'])
corr_coef_blr
```

```python
corr_coef_pune = compute_pearson_correlation(pune_df, ['LST'])
corr_coef_pune
```

```python
corr_coef_hyd = compute_pearson_correlation(hyd_df, ['LST'])
corr_coef_hyd
```

```python
corr_combined = pd.concat([corr_coef_blr, corr_coef_pune, corr_coef_hyd], axis=1)
corr_combined.columns = ['Bengaluru', 'Pune', 'Hyderabad']
corr_combined
```

```python
corr_combined = corr_combined.drop(index='Year')
corr_combined
```

```python
def plot_correlation_heatmap(df: pd.DataFrame, title: str = "Correlation Heatmap") -> None:
    """
    Plots a large, well-spaced heatmap for the given DataFrame using Plotly.

    :param df: A pandas DataFrame where rows are input variables and columns are cities or categories.
    :param title: Title of the heatmap.
    """
    fig = px.imshow(
        df,
        text_auto=".2f",
        color_continuous_scale="RdBu_r",
        title=title,
        labels=dict(x="Cities", y="Input Variables", color="Correlation"),
        aspect="auto"
    )

    fig.update_layout(
        width=1200,
        height=800,
        title_font_size=28,
        font=dict(size=16),
        margin=dict(t=80, b=80, l=100, r=100),
        xaxis=dict(tickfont=dict(size=18)),
        yaxis=dict(tickfont=dict(size=18), autorange="reversed"),
    )

    fig.show()
```

```python
plot_correlation_heatmap(corr_combined, title="Correlation Heatmap for Cities")
```
