### Calculating the magnitude of wind speed and direction

```python
def calculateWindSpeed(df : pd.DataFrame) -> pd.DataFrame:
    """
    Calculate wind speed from u and v components.
    """
    df['WS'] = np.sqrt(df['u_component_of_wind_10m']**2 + df['v_component_of_wind_10m']**2)
    return df
```

```python
def calculateWindDirection(df : pd.DataFrame) -> pd.DataFrame:
    """
    Calculate wind direction from u and v components.
    """
    df['WSD'] = np.arctan2(df['u_component_of_wind_10m'], df['v_component_of_wind_10m']) * 180 / np.pi
    df['WSD'] = (df['WSD'] + 360) % 360  # Normalize to [0, 360)
    return df
```

## Converting the units to Celsisus, metres/second, pascal to hPa

```python
def lstCorrect(df : pd.DataFrame) -> pd.DataFrame:
    """
    Corrects the LST values based on the RH values.
    """
    df['LST'] = df['LST'] * 0.1
    return df
```

```python
def toCelsius(df : pd.DataFrame, col: str) -> pd.DataFrame:
    """
    Converts Kelvin to Celsius.
    """
    df[col] = df[col] - 273.15
    return df
```

```python
def paTohPa(df : pd.DataFrame, col: str) -> pd.DataFrame:
    """
    Converts Pa to hPa.
    """
    df[col] = df[col] / 100
    return df
```

```python
blr_df = lstCorrect(blr_df)
hyd_df = lstCorrect(hyd_df)
pune_df = lstCorrect(pune_df)
```

```python
blr_df = toCelsius(blr_df, 'DPT')
hyd_df = toCelsius(hyd_df, 'DPT')
pune_df = toCelsius(pune_df, 'DPT')
```

```python
blr_df = paTohPa(blr_df, 'AP')
pune_df = paTohPa(pune_df, 'AP')
hyd_df = paTohPa(hyd_df, 'AP')
```
